# **PARTE 2 — Interpretación (Correlación de Pearson)**

## **Análisis conceptual**

**1. ¿Qué significa una correlación negativa en este contexto?**

Significa que cuando una variable aumenta, el abandono del cliente disminuye.

Existe una relación inversa.

Ejemplo:

Si aumenta la satisfacción, baja la probabilidad de abandono.

**2. Explica la relación entre: satisfaccion y abandono_cliente**

La correlación es -0,875, lo que indica una correlación negativa (fuerte).

Esto significa que clientes más satisfechos tienen menor probabilidad de abandonar.

Puede ser la variable más importante del análisis.

**3. Explica por qué una variable con correlación cercana a 0 no es útil.**

Porque no muestra relación lineal clara con la variable objetivo.

Si cambia esa variable, abandono_cliente casi no cambia.

Por eso aporta poca capacidad predictiva.

**4. ¿Por qué no debemos incluir variables altamente correlacionadas entre sí?**

Porque entregan información repetida.

Esto puede causar:

- multicolinealidad
- modelos menos estables
- dificultad para interpretar resultados

El modelo recibe casi la misma información dos veces.
